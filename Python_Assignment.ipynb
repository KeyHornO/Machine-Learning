### **Week 2 Assignment | Python and Computing Environment Set-up**

```python

room_temp_deg_F = 70  #This figure can be chamged to a more suitable room temp.

def convert_temp(temp): #This is the function that does the conversion
  temp_deg_F = temp * 9/5 + 32
  return temp_deg_F

option = "Y"

while (option == "Y"):  # This loop allows user to repeat conversion operations
  temp_deg_C = float(input("\nEnter temp. in °C >> "))
  temp_deg_F = convert_temp(temp_deg_C)

  if temp_deg_F > room_temp_deg_F:
    print("\n",temp_deg_C, "\b°C converts to ", temp_deg_F, "\b°F, which is above the typical room temperature of", room_temp_deg_F, "\b°F")
  else:
    print("\n",temp_deg_C, "\b°C converts to", temp_deg_F, "\b°F, which is below the typical room temperature of", room_temp_deg_F, "\b°F")

  option = input("\nEnter Y to continue, any other entry to exit >> ")

```
